```python
#Hyperparameter tuning using GridSearchCV
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

#Creating a model using RandomForestClassifier
rf = RandomForestClassifier(
    class_weight='balanced',
    random_state=42
)

#Defining a grid of hyperparameter
param_grid = {
    'n_estimators': [100, 300], #Number of trees in the forest
    'max_depth': [None, 10, 20], #Maximum depth of a tree
    'min_samples_split': [2, 10], #Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 5], #Minimum number of samples required to be at a leaf node
    'max_features': ['sqrt', 'log2'] #Number of features to consider when looking for the best split
}

#GridSearchCV will find the best combination of hyperparameters
grid_search = GridSearchCV(
    estimator=rf, #model to be runed
    param_grid=param_grid, #dictionary of hyperparameters
    scoring='roc_auc', #used to evaluate each combination
    cv=5, #5-fold cross-validation
    n_jobs=-1, #all available CPU cores
    verbose=2 #Prints detailed progress messages while fitting
)

#Training the model on data
grid_search.fit(X_train, Y_train)

#2x3x2x3x2=48x5=240 Total 240 models have been trained. For each 48 combinations of hyperparameters, data is splitted into 5 #parts. 
#Each part is 1 validation set while rest is training
```

```python
#Getting best combination of parameter
grid_search.best_params_
```

```python
#Getting best estimator
grid_search.best_estimator_
```

```python
#The best combination of hyperparameters is chosen based on mean ROC AUC across the 5 folds.
grid_search.best_score_
```

Summary:

Define a Random Forest model.
Tell the code which hyperparameters to try.
GridSearchCV tests all combinations of hyperparameters using 5-fold cross-validation.
It finds the combination that gives the best ROC AUC score.
Use this best model to make predictions on data.

```python
#Best model out of 240 models
best_rf = grid_search.best_estimator_
```

```python
#Evaluate Tuned Model on Test Set

from sklearn.metrics import classification_report, roc_auc_score

best_rf_pred = best_rf.predict(X_test)
best_rf_pred_proba = best_rf.predict_proba(X_test)[:,1]

print(classification_report(Y_test, best_rf_pred))
print('ROC_AUC:', roc_auc_score(Y_test, best_rf_pred_proba))
```

Recall: 76% for customers going to churn (good recall value!!) and 77% for non-churner, misses 23% of non-churners compared to 28% in LogisticRegression model

Overall accuracy: 77% (Better by 3% than LogisticRegression model)

```python
#Checking threshold over the range
import numpy as np

thresholds = np.arange(0.4, 0.9, 0.05)

for t in thresholds:
    preds = (best_rf_pred_proba > t).astype(int)
    print(f"Threshold {t}")
    print(classification_report(Y_test, preds))
#0.7 is giving highest accuracy of 80%
```
